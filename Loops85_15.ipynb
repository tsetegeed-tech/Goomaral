Бодлого Гараас өгсөн list доторх элементийг тоолдог код бич.

```python
arr = list(map(int, input().split()))

count_dict = {}

for el in arr:
  if el in cout_dict:
    cout_dict[el] += 1
  else:
    cout_dict[el] = 1

  print(count_dict)
```

```python
for k in count_dict:
  print('{} элемент {} удаа орсон байна.'.format(k, count_dict[k]))
```

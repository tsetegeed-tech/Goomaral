Бодлого 6_02. Багш нарын мэдээлэл 2

```python
# Алхам 1: input.txt файлыг үүсгэх (Жишээ оролтыг энд хуулж тавина)
input_data = """Уламбаяр Төртогтох эрэгтэй МБУС МЗТ 99044079
Баттогтох Цэен-Ойдов эрэгтэй МБУС МЗТ 99003771
Сандагдорж Балдорж эрэгтэй МБУС МТ 88859959
Уртнасан Гантогтох эмэгтэй МБУС МЗТ 88081699
Алимаа Бор эмэгтэй НХУС ТТ 88169994
Тунгалаг Балжир эмэгтэй НХУС АХТ 80159991"""

with open('input.txt', 'w', encoding='utf-8') as f:
    f.write(input_data)

# Алхам 2: Үүрэн холбооны операторыг тодорхойлох функц
def get_operator(phone):
    prefix = phone[:2]
    # Монголын үүрэн холбооны операторын угтварууд (жишиг байдлаар ангилав)
    if prefix in ['99', '94', '80', '85']:
        return 'Мобиком'
    elif prefix in ['88', '89', '86', '87']:
        return 'Юнител'
    elif prefix in ['91', '93', '98']:
        return 'Скайтел'
    elif prefix in ['90', '81', '82']:
        return 'Жимобайл'
    else:
        return 'Бусад'

# Файлаас унших
with open('input.txt', 'r', encoding='utf-8') as ifile:
    mylines = ifile.readlines()

operator_counts = {'Мобиком': 0, 'Юнител': 0, 'Скайтел': 0, 'Жимобайл': 0}

for line in mylines:
    values = line.split()
    if not values:
        continue
    phone = values[-1] # Сүүлийн элемент нь утасны дугаар байна
    op = get_operator(phone)
    if op in operator_counts:
        operator_counts[op] += 1

# Хамгийн олон хэрэглэгчтэй операторыг олох
max_op = max(operator_counts, key=operator_counts.get)
max_count = operator_counts[max_op]

# Гаралтын файлыг үүсгэх
with open('output.txt', 'w', encoding='utf-8') as ofile:
    ofile.write(f"{max_op} {max_count}\n")

# Үр дүнг хэвлэж шалгах
print("--- output.txt файлын агуулга ---")
with open('output.txt', 'r', encoding='utf-8') as f:
    print(f.read())
```

Бодлого 6_03: Бутархайг хураах


```python
# Хоёр тооны Хамгийн их ерөнхий хуваагч (ХИЕХ) олох рекурсив/энгийн функц
def hiekh(a, b):
    while b != 0:
        a, b = b, a % b
    return a

# Бутархайг хураах функц
def huraah(a, b):
    # ХИЕХ-ийг олох
    h = hiekh(a, b)

    # Хүртвэр болон хуваарийг ХИЕХ-д хувааж хураана
    h_urtver = a // h
    h_huvaari = b // h

    return h_urtver, h_huvaari

# Жишээ хэрэглээ: 24/36 бутархайг хураах
a = 24
b = 36

urtver, huvaari = huraah(a, b)
print(f"Анхны бутархай: {a}/{b}")
print(f"Хураасан бутархай: {urtver}/{huvaari}")
```

Бодлого 6_04: Фибоначийн тоон дараалал

```python
def fibonacci_check(n):
    if n < 0:
        return False, f"{n} нь натурал тоо биш байна."

    # 0 болон 1-р гишүүдийг шалгах
    if n == 0:
        return True, 0
    if n == 1:
        return True, 1 # эсвэл 2-р гишүүн (1 индекс өгөгдлөөс хамаарна)

    a, b = 0, 1
    index = 1

    while b < n:
        a, b = b, a + b
        index += 1

    if b == n:
        return True, index + 1
    else:
        return False, f"{n} нь Фибоначийн тоо биш байна."

# Жишээ хэрэглээ:
test_num = 8
is_fib, result = fibonacci_check(test_num)

if is_fib:
    print(f"{test_num} нь Фибоначийн тоо мөн бөгөөд дарааллын {result}-р гишүүн байна.")
else:
    print(result)
```

Бодлого 6_05: Тэгш өнцөгтийн класс

```python
class Rectangle:
    # Байгуулагч функц (Constructor) - 4 өөр хувилбараар утга авах боломжтой
    def __init__(self, a=0, b=0, name="Rectangle", source_type="param", source_file=None, other_obj=None):
        if source_type == "param":
            # 1. Параметрээс утга авах
            self.a = a
            self.b = b
            self.name = name

        elif source_type == "keyboard":
            # 2. Гараас унших
            self.name = input(f"Нэр оруулна уу: ")
            self.a = float(input(f"{self.name} тэгш өнцөгтийн урт (a) оруулна уу: "))
            self.b = float(input(f"{self.name} тэгш өнцөгтийн өргөн (b) оруулна уу: "))

        elif source_type == "file" and source_file:
            # 3. Файлаас утга авах (файлын эхний мөрөөс нэр, a, b-г уншина гэж үзнэ)
            try:
                with open(source_file, 'r', encoding='utf-8') as f:
                    line = f.readline().split()
                    if line:
                        self.name = line[0]
                        self.a = float(line[1])
                        self.b = float(line[2])
            except FileNotFoundError:
                self.name = name
                self.a = a
                self.b = b
                print(f"Анхааруулга: {source_file} файл олдсонгүй, анхдагч утгыг ашиглалаа.")

        elif source_type == "copy" and other_obj:
            # 4. Өөр объектоос хуулж авах
            self.a = other_obj.a
            self.b = other_obj.b
            self.name = name # Шинэ нэр өгөх эсвэл other_obj.name-г хуулж болно

    # Талбай олох гишүүн функц
    def get_area(self):
        return self.a * self.b

    # Периметр олох гишүүн функц
    def get_perimeter(self):
        return 2 * (self.a + self.b)

    # Мэдээллийг хэвлэх функц
    def display(self):
        print(f"Объект: {self.name} | Урт (a): {self.a} | Өргөн (b): {self.b} | Талбай: {self.get_area()} | Периметр: {self.get_perimeter()}")


# --- Дээрх 4 хувилбараар 4 объект үүсгэх жишээ ---

# Тестийн зорилгоор түр зуур файлд мэдээлэл бичиж бэлдэх
with open("rect_data.txt", "w", encoding='utf-8') as f:
    f.write("FileRect 7.5 3.2")

print("--- 4 өөр байгуулагчаар объект үүсгэх туршилт ---\n")

# 1. Параметрээс утга авдаг объект
rect1 = Rectangle(10, 5, "ParamRect", source_type="param")
rect1.display()

# 2. Файлаас утга авдаг объект
rect2 = Rectangle(source_type="file", source_file="rect_data.txt")
rect2.display()

# 3. Өөр объектоос хуулж авдаг объект (rect1-ийн утгыг хуулна)
rect3 = Rectangle(a=rect1.a, b=rect1.b, name="CopyRect", source_type="copy", other_obj=rect1)
rect3.display()

# 4. Гараас утга уншиж авдаг объект (Энд кодыг шууд ажиллуулахад зориулж коммент болгосон байна. Та гараас оруулах бол source_type="keyboard" гэж ашиглана)
# rect4 = Rectangle(source_type="keyboard")
# rect4.display()
```

Бодлого 6_06: Квадрат тэгшитгэлийн класс



```python
import cmath  # Сөрөг дискриминанттай үед комплекс шийд олоход ашиглана

class EqQuad:
    # Байгуулагч функц - 4 өөр хувилбараар утга авах боломжтой
    def __init__(self, a=0, b=0, c=0, source_type="param", source_file=None, other_obj=None):
        if source_type == "param":
            # 1. Параметрээс утга авах
            self.a = a
            self.b = b
            self.c = c

        elif source_type == "keyboard":
            # 2. Гараас унших
            print("Квадрат тэгшитгэлийн коэффициентүүдийг оруулна уу (ax^2 + bx + c = 0):")
            self.a = float(input("a = "))
            self.b = float(input("b = "))
            self.c = float(input("c = "))

        elif source_type == "file" and source_file:
            # 3. Файлаас утга авах (файлын эхний мөрөнд a, b, c зайгаар тусгаарлагдаж өгөгдсөн байна)
            try:
                with open(source_file, 'r', encoding='utf-8') as f:
                    line = f.readline().split()
                    if line:
                        self.a = float(line[0])
                        self.b = float(line[1])
                        self.c = float(line[2])
            except FileNotFoundError:
                self.a, self.b, self.c = a, b, c
                print(f"Анхааруулга: {source_file} файл олдсонгүй, анхдагч утгыг ашиглалаа.")

        elif source_type == "copy" and other_obj:
            # 4. Өөр объектоос хуулж авах
            self.a = other_obj.a
            self.b = other_obj.b
            self.c = other_obj.c

    # Тэгшитгэлийн шийд олох гишүүн функц
    def bodolt(self):
        print(f"Тэгшитгэл: {self.a}x^2 + {self.b}x + {self.c} = 0")

        if self.a == 0:
            print("Энэ нь квадрат тэгшитгэл биш байна (a = 0).")
            if self.b != 0:
                print(f"Шийд: x = {-self.c / self.b}")
            else:
                print("Шийдгүй эсвэл төгсгөлгүй олон шийдтэй.")
            return

        # Дискриминант тооцоолох (D = b^2 - 4ac)
        d = self.b**2 - 4 * self.a * self.c
        print(f"Дискриминант (D) = {d}")

        if d > 0:
            x1 = (-self.b + d**0.5) / (2 * self.a)
            x2 = (-self.b - d**0.5) / (2 * self.a)
            print(f"Тэгшитгэл 2 ялгаатай бодит шийдтэй: x1 = {x1}, x2 = {x2}")
        elif d == 0:
            x1 = -self.b / (2 * self.a)
            print(f"Тэгшитгэл 1 давхардсан шийдтэй: x1 = x2 = {x1}")
        else:
            # Сөрөг дискриминанттай үед комплекс шийд гарна
            x1 = (-self.b + cmath.sqrt(d)) / (2 * self.a)
            x2 = (-self.b - cmath.sqrt(d)) / (2 * self.a)
            print(f"Тэгшитгэл комплекс шийдтэй: x1 = {x1}, x2 = {x2}")
        print("-" * 40)


# --- Дээрх 4 хувилбараар 4 объект үүсгэж бодолтыг хийх ---

# Туршилтын зорилгоор утга авах файлийг үүсгэх
with open("eq_data.txt", "w", encoding='utf-8') as f:
    f.write("1 -3 2")  # x^2 - 3x + 2 = 0 (Шийд нь 1 ба 2)

print("--- Квадрат тэгшитгэлийн бодолтууд ---\n")

# 1. Параметрээс утга авдаг объект
eq1 = EqQuad(a=1, b=-5, c=6, source_type="param")
eq1.bodolt()

# 2. Файлаас утга авдаг объект
eq2 = EqQuad(source_type="file", source_file="eq_data.txt")
eq2.bodolt()

# 3. Өөр объектоос хуулж авдаг объект (eq1-ийн утгыг хуулна)
eq3 = EqQuad(source_type="copy", other_obj=eq1)
eq3.bodolt()

# 4. Гараас утга уншиж авдаг объект (Та гараас оруулах бол доорх комментыг авч ашиглана)
# eq4 = EqQuad(source_type="keyboard")
# eq4.bodolt()
```
